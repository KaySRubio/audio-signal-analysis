```python
import os
import sys
from pathlib import Path
import nemo.collections.asr as nemo_asr
```

```python
# have to modify where notebook looks for imports so it looks in the project root, not the notebook folder
os.getcwd()
sys.path.append(str(Path().resolve().parent))
print(sys.path)
```

```python

asr_model = nemo_asr.models.ASRModel.from_pretrained(model_name="nvidia/parakeet-tdt_ctc-110m")
```

```python
filename = 'cats_59s'
filepath = f'../data/audio/{filename}.wav'
```

```python
output = asr_model.transcribe([filepath])
print(output[0].text)
```

```python
asr_model_1b = nemo_asr.models.ASRModel.from_pretrained(model_name="nvidia/parakeet-tdt_ctc-1.1b")
```

```python
filename2 = 'bridge_music'
filepath2 = f'../data/audio/{filename2}.wav'
```

```python
output = asr_model_1b.transcribe([filepath2])
print(output[0].text)
```
