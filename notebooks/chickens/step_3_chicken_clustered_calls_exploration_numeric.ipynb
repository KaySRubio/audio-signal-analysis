```python
# import the data
filename = '2026-07-31_cluster_adult_analysis_output'
df = pd.read_csv(f"../../data/output/chickens/{filename}.csv")
```

```python
columns_to_keep = [
  'duration_ms',
  'time_to_next_call_ms',
  'min_freq',
  'max_freq',
  'max_energy_freq',
  'freq_change',
  'slope',
  'zcr',
  'spectral_centroid',
  'spectral_bandwidth',
  'cluster',
]

df = df[columns_to_keep]
df
```

```python

for col in columns_to_keep:
  df_meta = df.groupby("cluster")[col].agg(
      num_of_files="count",
      mean_duration="mean",
      max_duration="max",
      min_duration="min",
      std_duration="std"
  ).reset_index()
  print(f"\n{col}")
  display(df_meta)
```
